# Fuctions with Outputs

It Allows you to have an output once the function is completed.

def fucntion():
    result = 3*2
    return result

The output is given by the return keyword.then we can use it in variable the way we like.

```python
def format_name(f_name,l_name):
    """This is going to take a first name and last name 
    and format it to title case of the string.
    """
    return f"{f_name.title()}  {l_name.title()}"


f_name = input("What is the first name? ")
l_name = input("What is the last name? ")

formatted_string = (format_name(f_name,l_name))
print(formatted_string)
```
