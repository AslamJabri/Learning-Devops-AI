# Dictionaries
they allow us to group together and tag the information into two parts.
key: value
the syntax for the dictionary is 
dict1 = {key: value}

```python
dict = {
    "fruit" : "apple",
    "veg": "kale",
    "mixed": "tomato"
}

print(dict["fruit"])

dict["new"] = "modifying the data"
print(dict)

for key in dict:
    print(key)
```
