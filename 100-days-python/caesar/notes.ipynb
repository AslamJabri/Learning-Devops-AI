```python
def greet():
    print("Hello")
    print("What day is today?")
    
greet()

def new_greet(name):
    print(f"Hello {name}")
new_greet("linux")
```

# Positional vs Keyword argument
- Positional argument is the value which give in the function in the order
ex:
def name(name,location):
    print(name)
    print(location)
here the position and the order matters.

- Keyword argument
we are being more clear and specifing the parameter and value into the call function.

def name(name="hi",location="new")
