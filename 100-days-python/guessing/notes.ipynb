# Namespace and Scope

* **Namespace** is *where the variables live* (like a physical dictionary or telephone directory of names mapped to objects).
* **Scope** is *where you are allowed to look* to find those names.

---

## 1. What is a Namespace?

A **Namespace** is just a system Python uses to make sure that every single name you create is completely unique and doesn't conflict with other names. Think of it as a **system of family surnames**.



In your code:

* The **Global Namespace** has a slot: `enemies` $\rightarrow$ `1`.
* The **Local Namespace** inside `scope_enemies()` has a slot: `enemies` $\rightarrow$ `2`.

They live in entirely different dictionaries, which is why Python doesn't get confused or overwrite `1` with `2`.

---

## 2. The 4 Levels of Scope (The LEGB Rule)

When you look for a variable in Python, Python searches through namespaces in a very specific order. This is called the **LEGB Rule**. It always starts looking from the inside out:

### 1. **L**ocal Scope

Variables defined inside the current function you are running.

```python
def scope_enemies():
    enemies = 2  # <--- Local to this function
    print(enemies) 

```

### 2. **E**nclosing Scope

This only happens if you nest a function *inside* another function. The inner function can look "up" into the outer function's variables.

```python
def outer_function():
    enemies = 5  # Enclosing scope for the inner function
    
    def inner_function():
        print(enemies) # Python looks here second!

```

### 3. **G**lobal Scope

Variables defined at the very top level of your python file (exactly what you described).

```python
enemies = 1 # <--- Global Scope

```

### 4. **B**uilt-in Scope

These are the names that Python automatically loads for you the second you start up a script. Things like `print()`, `int()`, `len()`, or `input()`. You don't have to define them; they live in the outermost layer.

---

## Summarizing Your Example with LEGB

Let's look at how Python executes your code step-by-step based on these rules:

```python
enemies = 1 # Added to GLOBAL namespace

def scope_enemies():
    enemies = 2 # Added to LOCAL namespace
    print(enemies) # 1. Looks Local? YES. Prints 2.

scope_enemies()
print(enemies) # 1. Looks Local? No function here. 
               # 2. Looks Global? YES. Prints 1.

```

Because of **Scope**, the `print(enemies)` inside the function is trapped inside the **Local** wall, so it sees `2`. The `print(enemies)` at the bottom of the file is out in the **Global** space, so it only sees `1`.
