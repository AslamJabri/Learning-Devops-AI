## Data Types
Strings - group of characters is called strings. we have to create them with double quotes. Whatever comes in double quotes everything is treated as string only even if you put numbers as well.


```python
#Subscripting
print("Hello"[0])
print("Hello"[-1])
```

Integer - any number whole number is called integer.all the arthimetic operations can be done on the whole numbers.

```python
print(12+23)
```

Float - any number which has=ve a decimal point it is called float.

```python
print(3.14*2)
```

Boolean - it is a datatype which gives us if something is true or false

```python
print(True)
print(False)
```

TypeError: When function is expecting a string and we gave the integer value thats when we get type error

```python
len(123) # this will give type error
len("Hello") # As len function takes str
```

How to know what datatype is that we can use type(datatype)

```python
print(type(123))
print(type("abc"))
print(type(2.434))
print(type(True))
```

TypeCasting: This is something where we want to change the data type.

```python
a = "123" # it is a string here
print(type(a))
b = int(a) # here changed to int
print(type(b))
```

ValueError: it is something when the datatype cannot be converted into some other data type.
Like abc cant be converted into int

```python
print("Number of letters in your name: " + str(len(input("Enter your name: "))))
```

## Mathematical Operands

+ -> addition # 5+2 = 7
- -> substraction # 5-2= 3
* -> Multiplication # 5*2= 10
/ -> Division  # 5/2 = 2.5
// -> floor division  # 5//2 = 2
** -> Exponent / Power  # 5 ** 2 = 25
% -> modulo (Remainder) # 5 % 2 =1

```python
print(5+2)
print(5-2)
print(5*2)
print(5/2)
print(5//2)
print(5**2)
print(5%2)
```

When we have multiple operation on one line then what will be the highest order of calculation happens.
PEMDAS
P - Parentheses
E - Exponent
M or D - multiplication or division
+ or - - addition or subtraction

```python
print(3*3+3/3-3)
print(3*(3+3)/3-3)
```

## The Assignment Operator (Syntactic Sugar)

These are officially called Augmented Assignment Operators. They are huge time-savers because they let you modify a variable in place without re-typing its name.

You can use this pattern for almost all the operators we just discussed:

    score -= 1 (Decrement)

    score *= 2 (Double)

    score /= 2 (Halve)

    score = 2 (Square the current score)

The f-string (Literal String Interpolation)
 f-strings are the gold standard for Python string formatting. They are faster and more readable than the old .format() or %s methods.

```python
price = 15.5
    # Formatting to 2 decimal places (vital for money!)
print(f"The total is: ${price:.2f}") 
    # Output: The total is: $15.50
```
