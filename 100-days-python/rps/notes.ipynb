List
PYTHON LIST NOTES

- Ordered mutable collection
- Stores multiple values
- Uses square brackets []

Example:
fruits = ["apple", "banana"]

Properties:
- ordered
- mutable
- dynamic size
- allows duplicates

Indexing:
- starts from 0
- index represents memory offset

Common Operations:
append() -> add item at end
pop() -> remove and return item
insert() -> insert at specific position
remove() -> remove specific value

Important:
Variables store references to lists.

Example:
a = [1,2]
b = a

Both point to same object.

Performance:
append() -> usually O(1)
pop() from end -> O(1)
pop(0) -> O(n)
