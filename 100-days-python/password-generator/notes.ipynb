for loop is something where we want to repeat it to the definite value.The loop allow us to execute the same statement multiple times.

```python
fruits = ["apple","Peach","Pear"]

for fruit in fruits:
    print(fruit.title())
```

```python
student_score = [150,142,185,120,171,184,149,24,59,68,199,78,65,89,86]
num = 0
for score in student_score:
    if score > num:
        num = score
print(num)
```

range function in loop 
if we want to completely use the independent list without using the values we can use range.

```python
total = 0
for n in range(1,101):
    total += n
print(total)
```

```python
for i in range(1,101):
    if i%3==0:
        print("Fizz")
    elif i %5 ==0:
        print("Buzz")
    elif i % 3 ==0 and i%5==0:
        print("FizzBuzz")
    else:
        print(i)
```
